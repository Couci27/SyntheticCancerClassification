# Visualize Data

In order to understand the dataset and what the files contain


```python
import os
import pandas as pd
import numpy as np
```

```python
# Read pickled df
df = pd.read_pickle("merged_df.pkl")
df.head()
```

```python
# get the path to the first RNA parquet file
first_rna_path = df.loc[df["RNA-path"].notna(), "RNA-path"].iloc[0]

rna_data = pd.read_parquet(first_rna_path)


print(rna_data.shape)
print(rna_data.columns)
rna_data.head()
```

```python
# Get path to the first DNA-methylation parquet file
first_dna_path = df.loc[df["meth-path"].notna(), "meth-path"].iloc[0]

dna_data = pd.read_parquet(first_dna_path)

print(dna_data.shape)
print(dna_data.columns)
dna_data.head()
```

```python
# Get path to the first mirna parquet file
first_mirna_path = df.loc[df["mirna-path"].notna(), "mirna-path"].iloc[0]

mirna_data = pd.read_parquet(first_mirna_path)
print(mirna_data.shape)
print(mirna_data.columns)
mirna_data.head()
```
